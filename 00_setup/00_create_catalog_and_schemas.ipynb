```python
%sql
CREATE CATALOG IF NOT EXISTS banklens;

CREATE SCHEMA IF NOT EXISTS banklens.bronze;
CREATE SCHEMA IF NOT EXISTS banklens.silver;
CREATE SCHEMA IF NOT EXISTS banklens.gold;
CREATE SCHEMA IF NOT EXISTS banklens.data_quality;

CREATE VOLUME IF NOT EXISTS banklens.bronze.raw_files;
```

```python
%sql
-- Audit and control tables
-- Audit and control tables

CREATE TABLE IF NOT EXISTS banklens.data_quality.pipeline_audit_log (
    run_id STRING,
    pipeline_layer STRING,
    table_name STRING,
    day_number INT,
    source_path STRING,
    started_at TIMESTAMP,
    completed_at TIMESTAMP,
    source_record_count BIGINT,
    target_record_count BIGINT,
    status STRING,
    error_message STRING
);
```

```python
%sql
CREATE TABLE IF NOT EXISTS banklens.data_quality.pipeline_control (
    table_name STRING,
    pipeline_layer STRING,
    last_successful_day INT,
    last_run_at TIMESTAMP,
    total_runs INT
);
```

```python
%sql
CREATE TABLE IF NOT EXISTS banklens.data_quality.schema_change_log (
    table_name STRING,
    day_number INT,
    detected_at TIMESTAMP,
    previous_columns ARRAY<STRING>,
    current_columns ARRAY<STRING>,
    columns_added ARRAY<STRING>,
    columns_removed ARRAY<STRING>,
    columns_renamed ARRAY<STRING>
);
```
